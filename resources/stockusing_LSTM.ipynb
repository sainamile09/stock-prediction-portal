```python
data_training = pd.DataFrame(df.Close[0:int(len(df)*0.7)])
data_testing = pd.DataFrame(df.Close[int(len(df)*0.7):int(len(df))])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler(feature_range=(0,1))
scaler
```

```python
data_training_scaler = scaler.fit_transform(data_training)
```

```python
data_training_scaler
```

```python
x_train = []
y_train = []

for i in range(100,data_training_scaler.shape[0]):
    x_train.append([i-100,i])
    y_train.append([i,0])
```

```python
x_train,y_train = np.array(x_train),np.array(y_train)
```

```python
x_train
```

```python
y_train
```

```python
 from keras.models import Sequential
from keras.layers import Dense,LSTM,Input
```

```python
model = Sequential()
model.add(Input(shape=(100,1)))
model.add(LSTM(units=128,activation='tanh',return_sequences=True))
model.add(LSTM(units=64))
model.add(Dense(25))
model.add(Dense(1))
```

```python
model.compile(optimizer='adam',loss='mean_squared_error')
model.fit(x_train,y_train,epochs=50)
```

```python
model.summary()
```

```python
model.save('stockpredict.keras')
```

```python
past_100_days = data_training.tail(100)
past_100_days
```

```python
final_df = pd.concat([past_100_days,data_testing],ignore_index=True)
```

```python
final_data = scaler.fit_transform(final_df)
```

```python
x_test=[]
y_test=[]
for i in range(100,final_data.shape[0]):
    x_test.append(final_data[i-100:i])
    y_test.append(final_data[i,0])
```

```python
x_test,y_test = np.array(x_test),np.array(y_test)
```

```python
y_predicted = model.predict(x_test)
```

```python
y_predicted = scaler.inverse_transform(y_predicted.reshape(-1,1)).flatten()
y_test = scaler.inverse_transform(y_test.reshape(-1,1)).flatten()
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_test,'b',label='original price')
plt.plot(y_predicted,'r',label='predicted price')
plt.xlabel('days')
plt.ylabel('price')
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test,y_predicted)
mse
```

```python
rmse = np.sqrt(mse)
rmse
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_test,y_predicted)
r2
```
